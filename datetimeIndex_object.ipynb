### A collection of pandas timestamp (to store multiple date and time)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Example 
dt_index=pd.DatetimeIndex(['2023/1/1','2021/1/1','2022/1/1'])
dt_index
```

```python
# using python datetime object
import datetime as dt
x=pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
x
```

```python
# using pd.timestamps
x=pd.DatetimeIndex([pd.Timestamp(2023,1,1),pd.Timestamp(2022,1,1),pd.Timestamp(2021,1,1)])
x
```

```python
# Using datetimeindex as series index
a=pd.Series([1,2,3],index= dt_index)
a
```
