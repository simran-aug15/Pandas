### A collection of pandas timestamp (to store multiple date and time)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Example 
dt_index=pd.DatetimeIndex(['2023/1/1','2021/1/1','2022/1/1'])
dt_index
```

```python
# using python datetime object
import datetime as dt
x=pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
x
```

```python
# using pd.timestamps
x=pd.DatetimeIndex([pd.Timestamp(2023,1,1),pd.Timestamp(2022,1,1),pd.Timestamp(2021,1,1)])
x
```

```python
# Using datetimeindex as series index
a=pd.Series([1,2,3],index= dt_index)
a
```

### date_range_functions

```python
# generate daily date in given range
x=pd.date_range(start='2023/1/5',end='2023/1/31',freq='D')
x
```

```python
# alternate days in a given range 
x=pd.date_range(start='2023/1/5',end='2023/2/5',freq='2D')
x
```

```python
# B : Bussiness days : days from (Mon-Fri)
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='B')
x
```

```python
# W : one week per day : Thu
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='W-Thu')
x
```

```python
# H : Hourly data
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='h')
x
```

```python
# M : Month end data
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='ME')
x
```

```python
# H : Hourly data at 5H
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='5h')
x
```

```python
# MS : Month start data
x=pd.date_range(start='2023/1/5',end='2023/5/28',freq='MS')
x
```

```python
# A : Year end data
x=pd.date_range(start='2023/1/5',end='2029/2/28',freq='YE')
x
```
