### date_range_functions

```python
# generate daily date in given range
x=pd.date_range(start='2023/1/5',end='2023/1/31',freq='D')
x
```

```python
# alternate days in a given range 
x=pd.date_range(start='2023/1/5',end='2023/2/5',freq='2D')
x
```

```python
# B : Bussiness days : days from (Mon-Fri)
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='B')
x
```

```python
# W : one week per day : Thu
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='W-Thu')
x
```

```python
# H : Hourly data
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='h')
x
```

```python
# ME : Month end data
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='ME')
x
```

```python
# h : Hourly data at 5H
x=pd.date_range(start='2023/1/5',end='2023/2/28',freq='5h')
x
```

```python
# MS : Month start data
x=pd.date_range(start='2023/1/5',end='2023/5/28',freq='MS')
x
```

```python
# YE : Year end data
x=pd.date_range(start='2023/1/5',end='2029/2/28',freq='YE')
x
```

```python
# Using periods(no. of result)
x=pd.date_range(start='2023/1/5',periods=25,freq='D')
x
```

### to_datetime function

```python
# simple Series example
s=pd.Series(['2023/1/1','2022/8/1','2021/11/1'])
pd.to_datetime(s).dt.month_name()
```

```python
# with simple example u can fetch day year and month also
s=pd.Series(['2023/1/1','2005/8/15','2006/11/30'])
pd.to_datetime(s).dt.day_name()
```

```python
# with erroes
s=pd.Series(['2023/1/1','2022/1/1','2021/1/1'])
# converting it into a date time
x=pd.to_datetime(s,errors='coerce')
x
```

```python
# fetch month from s
pd.to_datetime(s).dt.month
```
