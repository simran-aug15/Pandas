```python
import numpy as np
import pandas as pd
```

```python
#import the dataset used in this 
courses=pd.read_csv('courses.csv')
students=pd.read_csv('students.csv')
nov=pd.read_csv('reg-month1.csv')
dec=pd.read_csv('reg-month2.csv')
matches=pd.read_csv('matches.csv')
delivery=pd.read_csv('deliveries (1).csv')
```

```python
#concat : We can vertically stack two datasets using this function
regs=pd.concat([nov,dec])
regs 
```

```python
#But we have one issue in above code that index get repeated so to solve this problem use ignore_index
regs=pd.concat([nov,dec],ignore_index=True)
regs
```

```python
#And if u want to differentiate between the data but not by index use this instead
multi_idx=pd.concat([nov,dec],keys=['Nov','Dec'])
multi_idx
```

```python
#Multiindex : Accessing only nov index
multi_idx.loc['Nov']
```

```python
# To access particular data
multi_idx.loc[('Nov',0)]
```
