### Wide format is where we have a single row for every data point with multiple columns to hold the values of various attributes (In simple words more columns less rows)   

### Long format is where for each data point we have as many rows as the number of attributes and each row contains the value of a particular  attribute for a given data points.(In simple words more rows and less columns)

### melt : This is use to convert Wide data into long data
### pivot : This is use to convert Long data into Wide data

```python
# import 
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# melt 
pd.DataFrame({'cse' : [120]})
pd.DataFrame({'cse':[120]}).melt()
```

```python
# melt another example:
pd.DataFrame({'cse':[120],'ece':[100],'mech':[150]}).melt(var_name='branch',value_name='num_students')
```

```python
# melt example
pd.DataFrame({
    'branch':['cse','cse_ai','cse_gen_ai','ece'],
    '2023':[100,210,130,213],
    '2024':[190,329,340,489],
    '2025':[290,870,652,832],
    '2026':[500,600,800,900]
} 
).melt(id_vars=['branch'],var_name='year',value_name='students') # id_vars = columns you want to keep unchanged (as identifiers) mean it just skip the column  or to change 
```

```python
# Real world example of melt :
death=pd.read_csv('time_series_covid19_deaths_global.csv')
confirm=pd.read_csv('time_series_covid19_confirmed_global.csv')
death.head() # Data is in wide format
confirm.head() # Data is in wide format
death=death.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='num_deaths')
Confirm=confirm.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='num_cases')
#merge
Confirm.merge(death,on=['Province/State','Country/Region','Lat','Long','date'])[['Country/Region','date','num_cases','num_deaths']]
```

### Pivot table : The pivot table takes simple column wise data input and groups the entries into a two-Dimensional table that provides a multidimensional summarized of the data 

```python
df=sns.load_dataset('tips')
df.head()
```

```python
df.groupby('sex')[['total_bill']].mean()
```

```python
# using groupby
df.groupby(['sex','smoker'])['total_bill'].mean().unstack()
```

```python
#  using Pivot table
df.pivot_table(index='sex',columns='smoker',values='total_bill')
```

```python
#aggfunc : using this you can easily use maths function on dataset It will apply on all the values u want to display
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum')
```

```python
# Multidimensional
df.pivot_table(index=['sex','smoker'],columns=['day','time'],values='total_bill')
```

```python
# also using aggfunc
df.pivot_table(index=['sex','smoker'],columns=['day','time'],aggfunc={'size':'mean','tip':'max','total_bill':'sum'})
```

```python
#margins 
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum',margins=True)
```
