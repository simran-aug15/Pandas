```python
# import dataset of titanic
df=pd.read_csv('titanic.csv')
df.head()
```

### Common Functions

```python
# lower/upper/capitalize/title

#upper() : Change the name in upper case latters
df['Name'].str.upper()
```

```python
# capitalize : It will capitalize the !st latter of name
df['Name'].str.capitalize()
```

```python
# title 
df['Name'].str.title()
```

```python
# lower : It convert the letters in lowercase letters
df['Name'].str.lower()
```

```python
# len : Find the length of name and longest name
df['Name'].str.len().max()
```

```python
df['Name'][df['Name'].str.len()==82].values[0]
```

```python
# strip : It will remove all the unnecessary spaces in the code
df['Name'].str.strip()
```

```python
#split : get
df['lastname']=df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title','firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df.head()
```
